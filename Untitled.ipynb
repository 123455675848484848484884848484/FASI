```python
import pandas as pd
import numpy as np

# Fonction de bruitage
def _add_noise_to_row(row, noise_std=0.05):
    numeric_fields = [
        "Age lors de l'intervention", 
        "IMC (Kg/m2)", 
        "SC (m2)", 
        "Durée de séjour total",
        "STRAIN VG GLS : %",
        "Surface valvulaire aortique indexée : cm²/m2",
        "masse VG g/m2",
        "DIAMETRE SOUS AO OU ANNEAU AO"
    ]
    for field in numeric_fields:
        try:
            original_str = str(row[field]).replace(',', '.')
            original_value = float(original_str)
            noise = np.random.normal(0, noise_std * original_value)
            new_value = round(original_value + noise, 2)
            row[field] = str(new_value).replace('.', ',') if ',' in str(row[field]) else new_value
        except Exception:
            continue
    row['Horodateur'] = pd.Timestamp.now().strftime('%m/%d/%Y %H:%M:%S')
    return row

# Fonction principale d’augmentation réaliste avec contrôle du ratio classe 1
def augment_data_realistically(df, target_size=60, ratio_class1=0.3, noise_std=0.05):
    current_size = len(df)
    needed = target_size - current_size
    if needed <= 0:
        return df.copy()

    n_class1 = int(ratio_class1 * needed)
    n_class0 = needed - n_class1

    class1_rows = df[df['Décès en post-opératoire'] == 1]
    class0_rows = df[df['Décès en post-opératoire'] == 0]

    # Sécurité : si pas assez de lignes pour échantillonner
    if len(class1_rows) == 0:
        print("⚠️ Aucun cas 1 trouvé. Tous les ajouts seront de classe 0.")
        n_class1 = 0
        n_class0 = needed

    augmented_rows = []

    # Générer classe 1
    for _ in range(n_class1):
        row = class1_rows.sample(n=1, replace=True).copy().iloc[0]
        row = _add_noise_to_row(row, noise_std)
        augmented_rows.append(row)

    # Générer classe 0
    for _ in range(n_class0):
        row = class0_rows.sample(n=1, replace=True).copy().iloc[0]
        row = _add_noise_to_row(row, noise_std)
        augmented_rows.append(row)

    df_augmented = pd.DataFrame(augmented_rows)
    df_final = pd.concat([df, df_augmented], ignore_index=True)
    return df_final

# === UTILISATION ===

# Charger les données
df = pd.read_csv("corrections1.csv")

# Appliquer l'augmentation : cible 60 lignes, avec 30 % de classe 1 dans les lignes ajoutées
df_augmented = augment_data_realistically(df, target_size=60, ratio_class1=0.3)

# Affichage de la répartition
print("Répartition après augmentation :")
print(df_augmented['Décès en post-opératoire'].value_counts())

# Sauvegarde
df_augmented.to_csv("corrections1_augmented.csv", index=False)
print(" Fichier sauvegardé : corrections1_augmented.csv")
```

```python
#  Imports
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import classification_report, confusion_matrix
from imblearn.over_sampling import RandomOverSampler
import seaborn as sns

# 1. Charger le fichier
df = pd.read_csv("corrections1_augmented.csv")

# 2. Supprimer les colonnes non pertinentes
df = df.drop(columns=['Horodateur', 'nom prénom', 'numero du dossier', 'date de l\'intervention'])

# 3. Encodage et conversion des champs
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].astype(str).str.replace(',', '.')
    try:
        df[col] = df[col].astype(float)
    except ValueError:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col])

# 4. Ajouter manuellement des cas de classe 1 si trop faibles
while df['Décès en post-opératoire'].sum() < 3:
    ligne = df[df['Décès en post-opératoire'] == 0].sample(1, random_state=np.random.randint(100)).copy()
    ligne['Décès en post-opératoire'] = 1
    df = pd.concat([df, ligne], ignore_index=True)

# 5. Séparation des features (X) et de la cible (y)
X = df.drop(columns=['Décès en post-opératoire'])
y = df['Décès en post-opératoire']

# 6. Split avec stratification
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y, random_state=42)

# 7. Normalisation
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 8. Rééquilibrage avec suréchantillonnage
ros = RandomOverSampler(random_state=42)
X_train_resampled, y_train_resampled = ros.fit_resample(X_train_scaled, y_train)

# 9. Modèle Arbre de Décision (Decision Tree)
clf = DecisionTreeClassifier(random_state=42, class_weight="balanced", max_depth=5)
clf.fit(X_train_resampled, y_train_resampled)

# 10. Évaluation
y_pred = clf.predict(X_test_scaled)

print(" Matrice de confusion :")
cm = confusion_matrix(y_test, y_pred)
print(cm)

print("\n Rapport de classification :")
print(classification_report(y_test, y_pred, zero_division=0))

# 11. Affichage graphique de la matrice de confusion
plt.figure(figsize=(5,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=["Pas de décès", "Décès"], yticklabels=["Pas de décès", "Décès"])
plt.title("Matrice de confusion")
plt.xlabel("Prédit")
plt.ylabel("Réel")
plt.show()

# 12. Visualisation de l'arbre de décision
plt.figure(figsize=(20, 8))
plot_tree(clf, filled=True, feature_names=X.columns, class_names=["Pas de décès", "Décès"], rounded=True)
plt.title("Arbre de Décision : prédiction du décès post-opératoire")
plt.show()
```
