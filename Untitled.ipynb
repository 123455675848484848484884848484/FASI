```python
import pandas as pd
import numpy as np

def augment_data_realistically(df, target_size=60, noise_std=0.05):
    current_size = len(df)
    if current_size >= target_size:
        print("Aucune augmentation nécessaire.")
        return df.copy()

    needed = target_size - current_size
    augmented_rows = []

    for _ in range(needed):
        # Sélectionner une ligne au hasard
        row = df.sample(n=1).copy().iloc[0]

        # Colonnes numériques à bruiter
        numeric_fields = [
            "Age lors de l'intervention", 
            "IMC (Kg/m2)", 
            "SC (m2)", 
            "Durée de séjour total",
            "STRAIN VG GLS : %",
            "Surface valvulaire aortique indexée : cm²/m2",
            "masse VG g/m2",
            "DIAMETRE SOUS AO OU ANNEAU AO"
        ]

        for field in numeric_fields:
            try:
                # Conversion virgule -> point si nécessaire
                original_str = str(row[field]).replace(',', '.')
                original_value = float(original_str)
                noise = np.random.normal(0, noise_std * original_value)
                new_value = round(original_value + noise, 2)

                # Remettre le format d'origine si virgule
                row[field] = str(new_value).replace('.', ',') if ',' in str(row[field]) else new_value
            except Exception as e:
                continue  # Sauter en cas de problème (valeurs nulles, texte...)

        # Modifier la date pour éviter les doublons exacts
        row['Horodateur'] = pd.Timestamp.now().strftime('%m/%d/%Y %H:%M:%S')
        augmented_rows.append(row)

    df_augmented = pd.DataFrame(augmented_rows)
    df_final = pd.concat([df, df_augmented], ignore_index=True)

    return df_final


# === UTILISATION ===

# 1. Charger le fichier d'origine
df = pd.read_csv("corrections1.csv")

# 2. Appliquer l'augmentation
df_augmented = augment_data_realistically(df, target_size=60)

# 3. Sauvegarder dans un nouveau fichier
df_augmented.to_csv("corrections1_augmented.csv", index=False)

print("✅ Dataset augmenté avec succès ! Fichier sauvegardé sous 'corrections1_augmented.csv'")
```
